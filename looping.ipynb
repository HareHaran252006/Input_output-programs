```python
# Python program to print multiplication table
n = int(input())
for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
#Count Digits in a Number
n=int(input())
count=0
if n==0:
  count=1
else:
  while n>0:
    n=n//10
    count=count+1
print(count)
```

```python
n=int(input())
reverse=0
while n>0:
  digit=n%10
  reverse=reverse*10+digit
  n=n//10
print(reverse)
```

```python
# Python program to find sum of digits

n = int(input())

total = 0

while n > 0:
    digit = n % 10
    total = total + digit
    n = n // 10

print(total)
```

```python
# Python program to print divisors of a number

n = int(input())

for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
```

```python
# Python program to check prime number
n = int(input())
count = 0
for i in range(1, n + 1):
    if n % i == 0:
        count = count + 1
if count == 2:
    print("Prime")
else:
    print("Not Prime")
```
