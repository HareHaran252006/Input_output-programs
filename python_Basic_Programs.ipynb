```python
# Python program to print the size of char, int, float and double

import sys

char = 'A'
integer = 10
floating = 10.5
double = 20.123456789

print("Size of char:", sys.getsizeof(char), "bytes")
print("Size of int:", sys.getsizeof(integer), "bytes")
print("Size of float:", sys.getsizeof(floating), "bytes")
print("Size of double:", sys.getsizeof(double), "bytes")
```
